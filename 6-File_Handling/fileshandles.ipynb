### The Role of the with Statement and Code Blocks
- The primary purpose of a context manager is to replace repetitive try...finally blocks, which are manually used to guarantee that a cleanup action (like closing a file) always runs. 
- The with statement creates a runtime context where a specific block of indented code is executed. 
- a context manager is an object used with the with statement to manage resources and operations within a specific code block, automatically handling setup and cleanup even if errors occur. 

### How Context Managers Work (The Protocol)
Any object that implements the context manager protocol can be used with a with statement. This protocol requires two special "dunder" methods: 
1. __enter__():
- Called when execution enters the with block.
- It performs the setup logic (e.g., opening a file, acquiring a lock).
- Its return value is assigned to the variable after the as keyword (if provided).
2. __exit__(self, exc_type, exc_value, traceback):
- Called automatically when execution leaves the with block, whether it exits normally or due to an exception.
- It performs the teardown or cleanup logic (e.g., closing the file, releasing the lock).
- It receives information about any exception raised within the block. If it returns a truthy value (e.g., True), the exception is suppressed; otherwise, the exception is re-raised by the with statement. 

```python
# Without a context manager (verbose, error-prone if an exception occurs)
file = open('demo.txt', 'w')
try:
    file.write('Hola!')
finally:
    file.close()

# With a context manager and the 'with' statement (cleaner, safer)
with open('demo.txt', 'w') as opened_file:
    opened_file.write('Hola!')
# The file is automatically closed upon exiting the indented block
```

```python
# Reading
file1 = "textsunai.txt"
f = open(file1) # by default moode is read: 'r'
print(f'Here is the File object that is created by assigning the open method into the F variable : {f = }')

# to access the file contents we have to use .read() method
contents = f.read()
print(contents)

# lets access using the with keyword
with open(file1) as filecont:
    contentf = filecont.read()
contentf

# here we are Storing the file data into the contents variable in one go, that seems not to be an memory efficient, 
# so lets use an for loop to get the data in a buffer manner without storing at one go!! 

with open(file1) as filecont:
    n = 1
    for line in filecont:
        # print(n,line) # Uncomment this line and see the difference, comment below line 
        print(n,line.rstrip("\n")) # using this as to avoid the new line character in the text file and avoid more spaces as python observes new line from document as \n
        n += 1

# Professional approach for large files
with open("large_data.txt",mode = "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip()) # strip() removes trailing newlines
```

    ========= ===============================================================
    Character Meaning
    --------- ---------------------------------------------------------------
    'r'       open for reading (default)
    'w'       open for writing, truncating the file first

    'a'       open for writing, appending to the end of the file if it exists

    The default mode is 'rt' (open for reading text). For binary random

# The Modern Professional Way: pathlib
- While os and open are classic, modern Python (3.4+) uses the pathlib module. it treats paths as objects rather than strings, making your code much cleaner.

### Useful pathlib Checkers
- If you aren't sure where you are, you can run these quick commands to inspect your environment:

- Path.cwd().exists(): Returns True (it always should!).

- Path.cwd().is_dir(): Confirms it's a folder.

- list(Path.cwd().iterdir()): Shows you every file inside your current folder.

```python
from pathlib import Path

# Create a path object
file_path = Path("directoryforos/config.txt")

# Check existence and read text easily
if file_path.exists():
    content = file_path.read_text()
    
# Write text without even needing a 'with' block for simple tasks
file_path.write_text("New configuration data")
```

Both **Path.read_text()** and **Path.write_text()** are high-level wrapper methods provided by the pathlib module. They automatically handle the entire lifecycle of the file operation behind the scenes:
- Automatic Open and Close: They open the file, perform the read or write operation, and instantly close it.
- Built-in Resource Management: Under the hood, they use a with block (context manager), ensuring that the file descriptor is safely closed even if an error occurs during the read or write.

When would you need to close a file?

You only need to worry about closing files (or using a with block) if you use standard lower-level file manipulation techniques, such as:
- Using the built-in open() function: f = open('file.txt', 'r') (requires f.close()).
- Using Path.open() explicitly: f = file_path.open() (requires f.close()).

```python
# Reading a file and count number of characters, lines and words in a file given

def count_ln_chr_str(filenaam):
    with open(filenaam, mode="r") as fl:
        lns = fl.readlines()
        line_ct = len(lns)
        word_ct = sum(len(w.split()) for w in lns) 
        char_ct = sum(len(ln.strip()) for ln in lns)
        ## The .split() method in Python is highly useful for breaking a string into a list of smaller substrings based on a specified delimiter.
        ## This is a fundamental operation for data processing and text manipulation tasks. 
        
        # word_ct = [w.split() for w in lns] 
        # char_ct = [ln for ln in lns]
        
    return line_ct,word_ct,char_ct

filenaam = 'Multilines.txt'
lines,words,char = count_ln_chr_str(filenaam)

print(lines,words,char)
```

### writing and reading a file
- The w+ mode in Python is used to open a file for both reading and writing. 
- If the file does not exist, it will be created. 
- If the file exists, its content is truncated (i.e., the file is overwritten).

```python
with open("ritenrd.txt",mode = "w+") as file:
    file.write("Hi the imput is via w+ mode in the code \n")
    file.write("tada!!! lets read it after creating and making the changes.")
    # Move the file cursor to the beginning
    file.seek(0) # reason behind this is to set the pointer to starting point so it can read the contents from the file
    # , if not set to 0 the cursor would be at the end of the last line created
    # Read the content of the file
    contentf = file.read()
    print(contentf)
```
