```python
## Nested Disctionaries
students={
    "student1":{"name":"Ironman","age":32},
    "student2":{"name":"Spidey","age":17}
}
print(students)

## Access nested dictionaries elementss [dict name][dict key]
print(students["student2"]["name"])
print(students["student2"]["age"])

students.items()
```

```python
# Iterating Over nested dictionaries

for stud,stud_inf in students.items():
    print(":)"*13)
    print(f'{stud}:{stud_inf}')
    
    for naam,vays in stud_inf.items():
        print(f'{naam}:{vays}')
    print("*******"*3)
```

# Dictionary Comprehension

- Dictionary comprehension in Python is a concise, single-line method for creating a new dictionary from an existing iterable (like a list, tuple, or another dictionary).
- It offers a more "Pythonic" and often more efficient alternative to traditional for loops. 
- Syntax:
    - The basic syntax of dictionary comprehension is enclosed in curly braces {}: 
   * {key_expression: value_expression for item in iterable [if condition]}

```python
{x:x**2 for x in range(4)} #printing values of squares for key values
```

```python
## Condition dictionary comprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)

odd = {x:x**2 for x in range(10) if x%2 != 0}
print(odd)
```

```python
## Practical Examples

## USe a dictionary to count he frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4,1,2,3,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)

```
