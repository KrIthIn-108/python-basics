# Sets in python
- Sets are a built-in data type in Python used to store collections of unique items. 
- They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. 
- Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
# Creating a set
my_set = {5,4,3,2,1,"Numbers",5,4,3,2,1} #unique val only
print(my_set)
print(type(my_set))

my_empty_set=set()
print(type(my_empty_set))

#typecasting
# list ---> set()
lst=[1,2,2,3,4,4,5]
set(lst)
```

```python
## Basics Sets Operation
## Adiing and Removing Elements
pandlu = {"Apple",'Banana','Watermelon','pineapple','potato','guava'}

# to add an element
pandlu.add("cherry")
print(f'Added an element:{pandlu}')

# to add multiple elements using *UPDATE* :---> It expects an iterable like lists
pandlu.update(['Orange','Mango'])
print("Updated fruits:",pandlu)

# to Remove from the Elements:  
# Removes the specified element from the set. If the element is not found, it raises a Keyerror

pandlu.remove('potato')
print(f"Remove the imposter veggi from Set: {pandlu}")

# if u dont want to raise an Key error we can use DISCARD()
# Removes the specified element if it is present. If the element is not in the set, it does nothing and does not raise an error.

pandlu.discard('potato') # thiugh potato removed it wont cause error

# clear() is used to remove all elements from the set
#pop(): Removes and returns an arbitrary (random) element from the set because sets are unordered.

u = pandlu.pop()
print(u) #this is completely random
```

```python
# Mazza ayega rey biru
# lets see Set operations

## Set Memebership test
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)

## MAthematical Operation
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

### Union
union_set=set1.union(set2)
print(union_set)

## Intersection
intersection_set=set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)

set1={1,2,3,4,5,6,"hehe set 1"}
set2={4,5,6,7,8,9,"hehe set 2"}

# Difference Elements in the first set but not the second
print(f'Set 1 - set 2 : {set1.difference(set2)}')
print(f'Set 2 - set 1 : {set2.difference(set1)}')

#Symmetric Difference Elements in either set, but not in both it will remove common elemets
print(f'Set 1 - set 2 : {set1.symmetric_difference(set2)}')

yio = {1,4,5,6}
jio = {2,3,7}
jio.isdisjoint(yio)

# issubset() or <=: Returns True if all elements of one set are present in the other.
# issuperset() or >=: Returns True if one set contains all elements of the other.
# isdisjoint(): Returns True if two sets have no common elements. 
## Sets Methods
st1={1,2,3,4,5}
st2={3,4,5}

## is subset
print(st1.issubset(st2))

print(st1.issuperset(st2))
# In place operations can also be done
# Python also provides methods to update a set in place (modifying the original set) using assignment operators: 
# 1. update() or |=: Add elements from another set (union update).
# 2. intersection_update() or &=: Keep only common elements (intersection update).
# 3. difference_update() or -=: Remove elements present in another set (difference update).
# 4. symmetric_difference_update() or ^=: Update with elements present in either set but not both. 
```

```python
### Counting Unique words in text

text="In this tutorial we are discussing about sets"
words=text.split()
print(type(words))
## convert list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```

Sets are a powerful and flexible data type in Python that provide a way to store collections of unique elements. They support various operations such as union, intersection, difference, and symmetric difference, which are useful for mathematical computations. Understanding how to use sets and their associated methods can help you write more efficient and clean Python code, especially when dealing with unique collections and membership tests.
