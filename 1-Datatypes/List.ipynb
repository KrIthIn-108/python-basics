### SLICING in Lists
- slicing in python works as [ inclusive value : exclusive value : step val ]

```python
brands = ["Suzuki","Maruti","Mahindra","Mercedez","BMW","Audi","Skoda"]

print(f'Length of brands: {len(brands)}')

print(brands[0:3]) # first 3 - 1 element i.e 3 elements
print(brands[3:-1]) #inclusive and ignoring last element
print(brands[-1:]) # last element
print(brands[::-1]) # reversing the list

print(brands[::2]) # step values of 2 frm beginning
print(brands[1::2]) # step values of 2 frm 1 index
print("Teda step mowa:",brands[::-2]) # step values of -2 frm -2 index

brand = brands.copy() # cpy a list from parent template
brand[1:]="cars ra baboi" #it will append each string into indices from index 1
print(f'Printing the copy from parents part and appened val:{brand}')

part_a = brands[:4]
part_b = brands[4:]
print(f'This is part a:{part_a} \nThis is part b:{part_b} \nThis is magic of both slicings {part_a + part_b}')

# edi kodiga intresting ga undi chudu 
# print(brands[15]) #out of index anni error throw chestundi edi, but if we use the same via indexing [:15] no error; "list index out of range"
print(brands[:15]) #pretty useful when you dont know the length of items in the list

print(brands[1:-1]) # ignores first and last list elements : ) ------
```

### Built in Methods:
- append,pop,insert,index,remove,count,sort,reverse,clear

```python
fruits = ["Apple","Banana","Watermelon","Orange","Guava","Berries","Sweetpotato","Mango"]

fruits.append("onion") #append()
print(fruits)

popped_item = fruits.pop() #pop()
print(popped_item)
print(fruits)

fruits.insert(len(fruits),"kiwi") # insert(index number ,item)
print(fruits)

guava = fruits.index("Guava") #index() returns index val
print("Guava location:",guava)

fruits.remove("Banana") #remove() element is case sensitive "banana" not acceptable 
print(fruits)

fruits.insert(2,"Mango")
c = fruits.count("Mango") #count() used to count the number of occurance of a certain element in the list
print("Count of mangoes:",c)
print(fruits)

fruits.sort() # sort() default sort is ascending; never save a .sort() list into a new variable coz it will return "None" value
print(fruits)

fruits.sort(reverse=True) # reverse=True now equal to sort in descending order
print(fruits)

fruits.reverse()
print("Reverses the whole list:",fruits) #reverse()

fruits.clear() # removes all elements from list
print(f'Clear() used to delete all elements from Fruits:-->{fruits}')

```

# ENUMERATE 
- The enumerate() function in Python is a built-in utility that adds a counter to any iterable (like a list, tuple, or string) and returns it as an enumerate object. This allows you to easily access both the index (counter) and the value of each item simultaneously, which is particularly useful when looping. 

- Its also like an helper functions for Iterables like: List,Tuple and Dict...
- Enum returns a tuple for the iterable items.

```python
fruits = ['apple', 'banana', 'cherry']

for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")

print("---------------enumerate with starting counter start = {valu}------")

for index, fruit in enumerate(fruits,start = 1):
    print(f"Index {index}: {fruit}")
```

# List Comprehension
- Basics Syntax [expression for item in iterable]

- with conditional logic [expression for item in iterable if condition]

- Nested List Comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
#Normal Iteration basic
lst=[]
for x in range(10):
    lst.append(x**2)
print(lst)

#using list comprehension
print("--------------------using list comprehension------------------")
# l = [x**2 for x in range(10)]
# print(l)
sqaure=[num**2 for num in range(10)]
print(sqaure)
```

```python
# List Comprehension with Condition
lst=[]
for i in range(10):
    if i%2==0:
        lst.append(i)
print(lst)
print("--------------------using list comprehension------------------")        
#with list conditional comprehesion 
even_numbers=[num for num in range(10) if num%2==0]
print(even_numbers)
```

```python
# Nested List Comphrension

lst1=[1,2,3,4]
lst2=['a','b','c']

pair=[[i,j] for i in lst1 for j in lst2] # nested list comphrension

print(pair)
```

```python
# List Comprehension with function calls
words = ["hello", "world", "python", "list", "comprehension"]
lengths = [len(word) for word in words]
print(lengths)  

naam = ['abc','xyz','lmk opl','pqr str','rtr','ktm'] # replaces space using list functions
refinement = [name.upper().replace(" ","-") for name in naam]
print(refinement)
singulu = [name for name in naam if " " not in name] # prints only elements who dont have spaces
print(singulu)
```

```python
lister = [2,4,5,6,7,8,97,17,18,137,57]
# This adds 1 for every number that matches the condition
oddnums = [i for i in lister if i % 2 != 0]
count = sum(1 for n in lister if n % 2 != 0)
print(f'Count of Odd numbers are {count} from odd number list {oddnums}')
print("see ya")
```

## 📝 List Comprehension Cheat Sheet

List comprehensions provide a concise way to create lists. Use this table as a quick reference for matching your logic requirements to the correct syntax structure.

| Use Case | Core Purpose | Syntax Structure | Example |
| :--- | :--- | :--- | :--- |
| **`for if`** | **Filter elements** (Shrink the list based on a condition) | `[x for x in items if condition]` | `[x for x in numbers if x % 2 == 0]` <br> *(Keeps only evens)* |
| **`for if else`** | **Transform elements** (Keep list size the same, change values) | `[x if condition else y for x in items]` | `['Pass' if score >= 50 else 'Fail' for score in scores]` |
| **`for if elif else`** | **Multi-stage transformation** (Chained conditional logic) | `[x if cond1 else y if cond2 else z for x in items]` | `['A' if s >= 90 else 'B' if s >= 80 else 'C' for s in scores]` |

> ⚠️ **The Pythonic Warning:** While nested `if-elif-else` comprehensions are syntactically valid, overusing them can turn your elegant one-liner into an unreadable monstrosity. If it takes more than 3 seconds for a peer to understand, break it out into a traditional loop or a helper function!

```python
numbers = [-1, 0, 2, -3, 5, 1, 2, 3, 4, 5, 6]

# Keeps the number IF it is even
evens = [num for num in numbers if num % 2 == 0]

print(evens)
# Output: [2, 4, 6]


# Transforms every element into a string label
labels = ["Even" if num % 2 == 0 else "Odd" for num in numbers]

print(labels)
# Output: ['Odd', 'Even', 'Odd', 'Even', 'Odd', 'Even']

# Simulates if-elif-else logic by chaining conditions
results = ["Positive" if num > 0 else "Negative" if num < 0 else "Zero" for num in numbers]

print(results)
# Output: ['Negative', 'Zero', 'Positive', 'Negative', 'Positive']

```
