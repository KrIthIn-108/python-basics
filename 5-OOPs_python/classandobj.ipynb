# Class and Objects in Python
Think of a Blueprint versus a House:  
- A **class** is the blueprint (the design), and an **object** is the actual house built from that design. You can build ten houses from one blueprint, but each house might have a different interior color.


- Classes and objects are the foundation of Object-Oriented Programming (OOP) in Python, allowing for the organization of code into reusable, modular structures that represent real-world entities. 
- __Class__: A blueprint or template for creating objects, defining their attributes (data) and methods (behavior).
- __Object__: A specific instance of a class, created at runtime with its own unique data (state). 

1. **Defining a Class:**
In Python, a class is created using the class keyword. By convention, class names use CamelCase. 

````python
class Dog:
    # Class Attribute (shared by all instances)
    species = "Canis familiaris"

    # Constructor method (initializes new objects)
    def __init__(self, name, age):
        # Instance Attributes (unique to each object)
        self.name = name
        self.age = age

    # Method (behavior)
    def bark(self):
        return f"{self.name} says Woof!"
````

2. **Creating Objects (Instantiation):**
- You create an object by calling the class as if it were a function, passing the necessary arguments defined in the __init__ method. 
python


````python
# Creating instances of the Dog class
dog1 = Dog("Buddy", 3)
dog2 = Dog("Lucy", 5)

# Accessing attributes
print(dog1.name)  # Output: Buddy
print(dog2.species) # Output: Canis familiaris

# Calling methods
print(dog1.bark()) # Output: Buddy says Woof!
````

# Key Concepts
1. __init__ Method: Known as the constructor, this method is automatically called when a new object is created to initialize its attributes.
    -  '_init_' is a special method in Python classes.
    - It is used to initialize object attributes.
    - '_init_' is automatically called when you create a new instance of a class.
    - It enables you to set up the object according to your specific requirements during the object's creation.

2. self Parameter: Represents the specific instance of the class being operated on. It must be the first parameter in any instance method.
3. Attributes: Variables that belong to the class or instance.
4. Instance Variables: Unique to each object (defined inside __init__ with self).
5. Class Variables: Shared by all instances (defined directly under the class line).
6. Methods: Functions defined inside a class that describe the behaviors of an object. 

## Important Concepts
- Encapsulation: Bundling data and methods together and restricting direct access to some components, typically using underscores (e.g., self._name or self.__name).
- Inheritance: Allowing a child class to inherit attributes and methods from a parent class.
- Polymorphism: Using a unified interface to represent different underlying forms (data types). 

## Benefits of Classes and Objects
- Code Reusability: Create a template once, and make many objects.
- Organization: Group related data and behaviors together.
- Modularity: Easier to maintain and debug code. 

## **SELF** <> KEYWORD
- self represents the instance of a class. 
- It allows you to access and modify variables, properties, and methods specific to the object being created or used. 
- self is not a reserved keyword in Python. It is a strong, universal convention used to identify the current object. 

### Key Aspects of self:

- Automatic Passing: When you call a method on an instance (e.g., object.method()), Python automatically passes the instance as the first argument.
- Explicit Definition: While passed automatically, it must be explicitly defined as the first parameter in instance methods and the __init__ constructor.
- Instance Specificity: It allows multiple instances of the same class to maintain their own unique data. **Instance variable**
- Invoking Instance Methods: It allows one method to call another method within the same class.
- Distinguishes Variables: It differentiates between instance variables (e.g., self.name) and local variables within a method. 

### Why self is Used
- Accessing Instance Data: It links attributes to specific objects.
- Invoking Other Methods: A method can call another method within the same class using self.method_name().
- Code Clarity: It makes it explicit that a method is operating on a specific object. 

````python
class Person:
    def __init__(self, name, age):                
        self.name = name  # Instance variable
        self.age = age    # Instance variable

    def greet(self):
        print(f"Hello, my name is {self.name}") # Accessing instance variable

# Creating objects
p1 = Person("Alice", 30)
p2 = Person("Bob", 25)

p1.greet()  # Output: Hello, my name is Alice
p2.greet()  # Output: Hello, my name is Bob
````

### 🏗️ The Analogy: The Video Game "Player" Class

If Player 1 falls into lava, the game needs to subtract health. But it shouldn't take health away from everyone in the game! It needs to take health away from **"itself."** In Python, `self` is the character pointing to their own chest and saying:

#### 2. The Name Tag
Imagine every character has a **"Property List"** stuck to their back:
* **Player 1's list:** `self.name = "Steve"`, `self.health = 10`

When the game runs the `jump()` command for Player 1, `self` acts like a **GPS tracker**. It tells the computer: *"Go to Player 1's list, find THEIR legs, and make THEM move."*

### 💡 The Memory Tip: The "Sticky Note" Trick
Here is the best way to remember it forever:

> **`self` is a Sticky Note.**
> When you build something from a class, Python slaps a sticky note on it that says **"MY."**

* `self.name` = **MY** name.
* `self.speed` = **MY** speed.
* `self.jump()` = **MY** way of jumping.

Without that **"MY"** (the `self`), the computer gets confused and doesn't know whose name or speed you are talking about!

- self appears inside def __init__(self, ...):.

# 🐾 Understanding `self` with the Pet Blueprint

### The Breakdown: The Pet Blueprint
We have a blueprint for a **Pet**. Every pet has a **name** and a **sound**. But since every pet is different, we use `self` to make sure the Dog doesn't meow and the Cat doesn't bark.

### The Solution: "My" Name and "My" Sound
Notice how in the code below, `self` is like the pet pointing to its own name tag.

```python
class Pet:
    def __init__(self, name: str, sound: str):
        # Using the "Sticky Note" to remember MY details
        self.name = name
        self.sound = sound

    def speak(self):
        # Without 'self', the computer wouldn't know WHICH name to print!
        print(f"{self.name} says: {self.sound}!")

# Creating two different "Instances" (Objects)
dog = Pet(name="Buddy", sound="Woof")
cat = Pet(name="Whiskers", sound="Meow")

# Calling the function
dog.speak()  # Output: Buddy says: Woof!
cat.speak()  # Output: Whiskers says: Meow!

````python
class Superhero:
    # 1. Class Variable (Shared by EVERYONE)
    secret_identity = "Thanos" 

    def __init__(self, secret_identity):
        # 2. Instance Variable (Unique to THIS person)
        self.secret_identity = secret_identity

    def reveal(self):
        # Option A: To show "Peter Parker", use self
        print(f"I am actually {self.secret_identity}!")
        
        # Option B: To show "Thanos", use the Class Name
        print(f"But the Master Villain is {Superhero.secret_identity}!") # here we are accesing the class variable

spider_man = Superhero("Peter Parker")
spider_man.reveal()

# We use Superhero.secret_identity (without self) for things that are always true for every object made from that class. 
# For example, if you had a class Human, a class variable might be number_of_legs = 2. It doesn't change from person to person.
````

# 🏗️ Is `__init__` Compulsory?

Short answer: **No.** But it is the most common tool in a Pythonista's toolkit. Here is why:

### 1. The "Ghost" Constructor
If you don't write an `__init__` method, Python doesn't leave the class empty. It automatically provides a **default constructor** that does nothing. It's essentially invisible:

```python
def __init__(self):
    pass
```

### 2. When Can You Skip It?
You can skip `__init__` if your class doesn't need to store any unique data when it is "born." This is common in **Utility Classes** where the behavior is the same every time.

**Example: A Utility Class**
```python
class MathTools:
    def add(self, a, b):
        return a + b

# This works perfectly without an __init__ because the 
# calculator doesn't need to 'remember' anything about itself.
calculator = MathTools()
print(calculator.add(5, 10))
```

### 3. Why Is It Usually Necessary?
Without `__init__`, your objects are like **blank slates**. If you want your objects to have different names, ages, or settings from the moment they are created, you **must** use `__init__`.

---

### 📊 The Breakdown: With vs. Without `__init__`

| Feature | Without `__init__` | With `__init__` |
| :--- | :--- | :--- |
| **Setup** | Objects start "empty." | Objects start with data (name, health, etc.). |
| **Arguments** | You cannot pass data: `p = Person()` | You can pass data: `p = Person("Pythesh")` |
| **Usage** | Best for simple tools or math logic. | Best for real-world objects (Users, Products). |

````python
# Lets create a banking app as usually
from datetime import date,datetime
class banku:
       
    def __init__(self,actno,name,balance = 0): # default keyword argument balance set to 0
        self.actno = actno  # instance attributes/variables
        self.name = name
        self.balance = balance
          
    def balanceentha(self):
        return self.balance
    def deposit(self,amount):
        self.balance += amount
        print(f'Updated Balance for {self.name} is: {self.balance}')
        self.status()
    
    def withdraw(self,amount):
        if self.balance > amount:
            self.balance -= amount
            print(f'With draw amount {amount}, Updated Balance for {self.name} is: {self.balance} ')
        else:
            print(f'orey ne bank la balance levu, balance is {self.balance}')
        self.status()

    def status(self):
        print(f'{self.name} with Account number bearing {self.actno} has a final balance {self.balance} as of {datetime.now().strftime("%B %d, %Y at %I:%M %p")}')
    
ram = banku(123,"Ram",500)
ravi = banku(754,"Ravi",752)

print(ram.name)
print(ram.actno) # instance variables/attributes

ram.deposit(5000) #instance methods

ram.withdraw(875)
print("***************************************")
ravi.deposit(7000)
ravi.withdraw(45220)
# ravi.status() ## edo gelika
ram.balanceentha()
print(ravi.balanceentha())
````

### 🔄 Calling Methods Within a Class
To call one method from another inside the same class, you must use the `self` prefix. This tells Python to look for the method within the current object's scope.

**Syntax:** `self.method_name()`

**Example in Bank Class:**
In the code below, `self.status()` is called at the end of both `deposit` and `withdraw` to provide an immediate update to the user.

```python
def deposit(self, amount):
    self.balance += amount
    self.status()  # Correct way to call another instance method

# Magic Methods
- Magic methods in Python, also known as dunder methods (double underscore methods), are special methods that start and end with double underscores. These methods enable you to define the behavior of objects for built-in operations, such as arithmetic operations, comparisons, and more.

- Magic Methods
Magic methods are predefined methods in Python that you can override to change the behavior of your objects. Some common magic methods include:


1. __init__': Initializes a new instance of a class.
2. __str__: Returns a string representation of an object.
3. __repr__: Returns an official string representation of an object.
4. __len__: Returns the length of an object.
5. __getitem__: Gets an item from a container.
6. __setitem__: Sets an item in a container.
'''

````python
## Basics MEthods
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def __str__(self):
        return f"{self.name},{self.age} years old"
    
    def __repr__(self):
        return f"Person(name={self.name},age={self.age})"
    
person=Person("Mr Unknow",23)
print(person)
print(repr(person))
````
