# Inheritance in Python
- Single Inheritance : A Class Inherits from a single base class using **super().__init__()**
- Multiple Inheritance: A Class can Inherit Multiple Parent class using **classname.__init__()**

# 🧬 Inheritance: The "Gift of the Parent"

Inheritance allows you to build a specialized version of a class (Child) without rewriting the logic already present in an existing class (Parent).

### 1. The Definition
**Inheritance** is a mechanism where a new class (Subclass/Child) derives properties and behaviors (attributes and methods) from an existing class (Superclass/Parent).

* **Parent Class (Base):** The original blueprint containing general features.
* **Child Class (Derived):** The specialized version that "inherits" those features and adds its own unique logic.

---

### 2. Types of Inheritance in Python
Python supports several ways to pass down logic across classes:

| Type | Description | Analogy |
| :--- | :--- | :--- |
| **Single** | One Child inherits from One Parent. | Inheriting a house from a father. |
| **Multiple** | One Child inherits from Two or more Parents. | Inheriting eye color from Mom and height from Dad. |
| **Multilevel** | A Child inherits from a Parent, who inherited from a Grandparent. | A family lineage (Grandfather → Father → Son). |
| **Hierarchical** | Multiple Children inherit from the same single Parent. | Two siblings inheriting the same last name. |

---

### 3. How to Inherit? (Syntax & Code)
To inherit, place the Parent's name in parentheses after the Child's class name. We use `super()` to trigger the Parent's initialization logic.

```python
# The Parent (Base Class)
class Smartphone:
    def __init__(self, brand: str):
        self.brand = brand

    def call(self, number: str):
        print(f"Dialing {number} from your {self.brand}...")

# The Child (Derived Class)
class SmartCameraPhone(Smartphone): 
    def __init__(self, brand: str, megapixels: int):
        # super() triggers the Parent's __init__ to set the brand
        super().__init__(brand) 
        self.megapixels = megapixels

    def take_photo(self):
        print(f"Capturing a {self.megapixels}MP photo on your {self.brand}!")

# Usage
my_phone = SmartCameraPhone("Samsung", 108)
my_phone.call("911")       # Inherited method from Smartphone
my_phone.take_photo()      # Unique method in SmartCameraPhone

```
### 4. The "Golden Rules" of Inheritance
1. The super() Keyword: Use super().__init__() if your child class has its own __init__. If skipped, parent attributes (like self.brand) won't be initialized!

2. **Method Overriding**: If the Child has a method with the same name as the Parent, the Child's version "wins."

3. "Is-A" Relationship: Use inheritance only if you can say "The Child IS A Parent" (e.g., a Dog is an Animal).

4. Accessing Variables: Variables starting with __ (private) are not directly accessible by the Child. Use a single underscore _ if the Child needs access.

````python
class Grandparent:
    gold = "10kg"

class Parent(Grandparent):
    house = "Villa"

class Child(Parent):
    car = "Tesla"

me = Child()
print(f'Inherited Gold values is : {me.gold}')
````

````python
## Single Inheritance
## Inheritance (Single Inheritance)
## Parent class
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car ")

cr = Car(5,5,"disel v8")
print(cr.windows)
cr.drive()

# Lets inherit the car class and create another instance with inheritance using the derived class thar

class Thar(Car):
    def __init__(self,windows,doors,enginetype,crtype):
        super().__init__(windows,doors,enginetype)
        self.crtype = crtype
    def isitbyfour(self):
        print(f"This Car is a Thar and it is a {self.crtype}")
th = Thar(3,3,"stallion 1.2 lt","4X4")
th.drive()
th.isitbyfour()
````

````python
### Multiple Inheritance
## When a class inherits from more than one base class.
## Base class 1
class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        print("Subclass must implement this method")

## Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner


##Derived class
class Dog(Animal,Pet,Car): # here we are inheriting 3 classes
    def __init__(self,name,owner,windows,doors,enginetype):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
        Car.__init__(self,windows,doors,enginetype)
    def drive(self):
        print(f"{self.name}'s owner {self.owner} drives a {self.enginetype}")
    def speak(self):
        return f"{self.name} says wooffff"
    

## Create an object
dog=Dog("Gymson","K",3,3,"Thar")
dog.drive()
print(f'{dog.name} has a car with {dog.windows}')
print(dog.speak())
print(f"Owner is :{dog.owner}")
````

## Method Overiding and Method Overloading
### 1. Method Overriding (The "Upgrade")
 - Definition: This happens in Inheritance. It is when a Child class provides a specific implementation of a method that is already provided by its Parent class.
 - Goal: To change or "upgrade" the behavior of a parent's method.
 - Requirement: Must have a Parent-Child relationship.

 ``` python 
 class Bird:
    def move(self):
        print("The bird flies in the sky.")

class Penguin(Bird):
    # OVERRIDING: We change the move method because penguins can't fly
    def move(self):
        print("The penguin waddles on the ice.")

# Testing Overriding
generic_bird = Bird()
pingu = Penguin()

generic_bird.move() # Output: The bird flies in the sky.
pingu.move()        # Output: The penguin waddles on the ice.
 
 ```

 ### 2. Method Overloading (The "Multi-Tool")
   - Definition: In other languages, this means having multiple functions with the same name but different parameters. 
   - Python does not support traditional method overloading. If you define two functions with the same name, the second one will simply overwrite the first.
   - The Pythonic Way: We achieve "overloading" by using Default Arguments or *args.

``` python 
class Calculator:
    # We simulate overloading using default values (None)
    def add(self, a, b, c=None):
        if c is not None:
            return a + b + c
        else:
            return a + b

# Testing simulated Overloading
calc = Calculator()
print(calc.add(10, 20))      # Works with 2 arguments
print(calc.add(10, 20, 30))  # Works with 3 arguments

```

## Things to Remember
 * Overriding uses super(): If you want to use the parent's logic and add new logic, use super().method_name() inside the child's overridden method.

 * Overloading trick: If you need true complex overloading, you can use the @functools.singledispatch decorator, but for beginners, Default Arguments (like c=None) are the standard way.

 

````python
class Employee:
    """The Parent Class"""
    def __init__(self, name: str, salary: int):
        self.name = name
        self.salary = salary

    def work(self):
        print(f"{self.name} is now working...")

class Developer(Employee):
    """The Child Class"""
    def __init__(self, name: str, salary: int, language: str):
        # super() sends name and salary to the Parent's __init__
        super().__init__(name, salary)
        self.language = language  # Unique skill for Developer

    def work(self):
        # OVERRIDING the parent method
        print(f"{self.name} is coding in {self.language}!")

class Manager(Employee):
    """Another Child Class"""
    def schedule_meeting(self):
        print(f"{self.name} is scheduling a meeting.")

# --- Usage ---
dev = Developer("Pythesh", 80000, "Python")
mgr = Manager("Alice", 95000)

dev.work()           # Uses the overridden version: "Pythesh is coding..."
mgr.work()           # Uses the Parent version: "Alice is now working..."
mgr.schedule_meeting() # Unique to Manager
````
