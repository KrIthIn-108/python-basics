```python
## High level operations on files and collection of files
import shutil
shutil.copyfile('source.txt','destination.txt')
# copies value from source.txt to destination.txt
```

```python
## Data Serialization
import json
data={'name':'Mr Unknow','Age':22}

# Dict to json(string)
json_str=json.dumps(data)
print(json_str)
print(type(json_str))

# converting into json(string) to dict
parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv

import csv

with open('example.csv',mode='w',newline='') as file: # context managers csv file is also created
    writer=csv.writer(file)
    writer.writerow(['Name','Age','Grade'])
    writer.writerow(['Unknow',22,'A'])
    writer.writerow(['Know',22,'B'])
    
with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)

print(yesterday)
```
