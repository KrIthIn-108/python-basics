# Exception Handling 
- Exception handling in Python is a mechanism used to manage runtime errors, known as exceptions, allowing a program to continue running smoothly instead of crashing. It separates error-handling code from the main program logic, making code more robust and readable. 

## The core components of exception handling are the try, except, else, and finally blocks. 
### Key Concepts and Syntax
1. **try block**: This block contains the code that might raise an exception.
2. **except block**: This block catches and handles the exception if one occurs within the try block. You can specify the type of exception to catch, or a generic handler for all exceptions.
3. **else block**: The code in this optional block executes only if the code in the try block runs without raising any exceptions.
4. **finally block**: This optional block of code always executes, regardless of whether an exception occurred or was handled. It is commonly used for cleanup actions, such as closing files or releasing resources. 

- Exceptions are events that disrupt the normal flow of a program. They occur when an error is encountered during program execution. Common exceptions include:

1. ZeroDivisionError: Dividing by zero.
2. FileNotFoundError: File not found.
3. ValueError: Invalid value.
4. TypeError: Invalid type.

```python
#This example demonstrates how to use the try, except, else, and finally blocks to handle a common error: division by zero. 
try:
    # Code that may cause an exception
    numerator = 10
    denominator = int(input("Enter a number: "))
    result = numerator / denominator
except ZeroDivisionError:
    # Handle the specific error (division by zero)
    print("Error: You cannot divide by zero!")
except ValueError as r:
    # Handle another specific error (invalid input type)
    print("Error: Invalid input. Please enter an integer.",r)
else:
    # Code to execute if no exception occurs
    print(f"The result is {result}")
finally:
    # Code that always runs (cleanup, etc.)
    print("Execution completed.")
```

## Catching Specific and Multiple Exceptions
- It is a best practice to catch specific exceptions rather than using a broad, catch-all except statement, which can hide bugs. 
- Multiple except blocks: Use separate except blocks to handle different types of exceptions in distinct ways.
- Catching multiple exceptions in one block: You can group multiple exceptions into a single except block using a tuple if they require the same handling logic:

```python
try:
    a = 5
    b = '5'
    sum = a+b #throws a type error
    s = int(input("Enter a Number:"))
    
except (ValueError, TypeError) as e:
    print(f"An error occurred: {e}")
```

# Raising Custom Exceptions
- We can force an exception to occur using the raise keyword, which is useful for enforcing specific rules in our program: 

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.") # raise key word

    print("Access granted.")

try:
    a = int(input("Enter a ur Age:"))
    check_age(a)
except Exception as e:
    print(f"Caught the error: {e}")
```
